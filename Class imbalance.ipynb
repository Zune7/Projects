```python
import pandas as pd

# 1️⃣ Load dataset (change path if needed)
# Example:
dataset = pd.read_excel(r"D:\Thesis\Main\Data\GW table\Weekly_Groundwater_2023_Recharge class.xlsx")
# Assuming 'Recharge_Index' is the column name
print(dataset['Recharge_Index'].describe())

# Check how many would fall into your 5 classes
bins = [0, 0.10, 0.20, 0.30, 0.40, 1.0]
labels = [1, 2, 3, 4, 5]
temp_split = pd.cut(dataset['Recharge_Index'], bins=bins, labels=labels)
print(temp_split.value_counts())
```
