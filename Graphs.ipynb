```python
import matplotlib.pyplot as plt
import os

# Ensure the save directory exists
output_dir = "thesis_pie_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Structured Dataset with your exact GIS calculated percentages
factors_data = {
    "Drainage Density": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5"],
        "percentages": [35.29, 22.58, 21.29, 15.39, 5.45]
    },
    "Geology": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5", "Class 6", "Class 7"],
        "percentages": [11.31, 33.51, 10.25, 25.97, 15.06, 2.63, 1.26]
    },
    "Lineament Density": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5"],
        "percentages": [76.43, 6.89, 8.01, 6.67, 2.00]
    },
    "LULC": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4"],
        "percentages": [10.02, 79.63, 6.92, 3.43]
    },
    "Rainfall": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5"],
        "percentages": [9.68, 13.73, 25.02, 23.06, 28.50]
    },
    "Soil": {
        "labels": ["Class 1", "Class 2"],
        "percentages": [38.58, 61.42]
    },
    "TWI": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5"],
        "percentages": [42.03, 21.58, 20.20, 13.50, 2.69]
    },
    "Slope": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5"],
        "percentages": [35.24, 36.67, 22.04, 5.73, 0.32]
    },
    "DEM": {
        "labels": ["Class 1", "Class 2", "Class 3", "Class 4", "Class 5"],
        "percentages": [17.92, 39.87, 26.27, 8.18, 7.77]
    }
}

# 2. Batch Processing Loop
for factor_name, data in factors_data.items():
    labels = data["labels"]
    percentages = data["percentages"]
    
    # Dynamically pick palettes based on number of classes to maintain contrast
    if len(labels) > 5:
        colors = plt.cm.get_cmap('Set3')(range(len(labels)))
    else:
        colors = plt.cm.get_cmap('Pastel1')(range(len(labels)))
        
    # Configure high-quality figure canvas
    fig, ax = plt.subplots(figsize=(6, 5), subplot_kw=dict(aspect="equal"))
    
    # Generate the pie chart
    wedges, texts, autotexts = ax.pie(
        percentages, 
        labels=labels, 
        autopct='%1.1f%%',
        startangle=90,
        colors=colors,
        pctdistance=0.75, # Keeps numbers cleanly inside or nicely separated
        textprops=dict(fontname="Palatino Linotype", color="black"),
        wedgeprops=dict(width=1.0, edgecolor='white', linewidth=1.5) # Clean white grid dividers
    )
    
    # Style formatting rules
    plt.setp(autotexts, size=10, weight="bold")
    plt.setp(texts, size=11)
    
    # Set localized descriptive title
    ax.set_title(f"{factor_name} Spatial Distribution", fontname="Palatino Linotype", fontsize=14, weight="bold", pad=20)
    
    # Prevent tight packing layout issues
    plt.tight_layout()
    
    # Save chart at 300 DPI for ultra-sharp rendering in Word/LaTeX
    safe_filename = factor_name.lower().replace(" ", "_")
    save_path = os.path.join(output_dir, f"{safe_filename}_pie_chart.png")
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.close()

print(f"Success! All 9 high-quality charts have been saved inside the '{output_dir}' directory.")
```
