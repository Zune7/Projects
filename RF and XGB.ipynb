```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import xgboost as xgb
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix,
                             roc_auc_score, roc_curve, precision_recall_curve)
from sklearn.preprocessing import LabelEncoder
import joblib

# Load dataset
df = pd.read_excel(r'D:\Thesis\Main\Data\GW table\Dataset_binary11.xlsx')

# Define features and target
feature_cols = ['Rainfall', 'Soil', 'Geology', 'DEM', 'Drainage_Density',
                'Slope', 'Lineament_Density', 'LULC', 'TWI']
target_col = 'Recharge_Class'

X = df[feature_cols].values
y = df[target_col].values
```

```python
rf_final = RandomForestClassifier(
    n_estimators=90,
    max_depth=10,
    min_samples_split=24,
    min_samples_leaf=14,
    max_features='sqrt',
    bootstrap=True,
    oob_score=True,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_final.fit(X, y)

# Save model
joblib.dump(rf_final, r'D:\Thesis\Main\Models\Random Forest\New Final\RF_Final_Model11.pkl')
print("✓ Final model saved: RF_Final_Model11.pkl")

# Save feature names for later use
joblib.dump(feature_cols, r'D:\Thesis\Main\Models\Random Forest\New Final\RF_Final_Model_Features_name11.pkl')
print("✓ Feature names saved: feature_names11.pkl")
```

```python
xgb_cv = xgb.XGBClassifier(
    n_estimators=30,
    max_depth=4,
    learning_rate=0.05,
    subsample=1,
    colsample_bytree=1,
    min_child_weight=1,
    gamma=0,
    reg_alpha=0,
    reg_lambda=1,
    objective='binary:logistic',  # correct for binary classification
    random_state=42,
    n_jobs=-1
)


xgb_cv.fit(X, y)
joblib.dump(xgb_cv, r'D:\Thesis\Main\Models\XGBoost\XGBoost_KFold_Model111.pkl')
print("✓ XGBoost K-Fold model saved: XGBoost_KFold_Model11.pkl")


# Save feature names for later use
joblib.dump(feature_cols, r'D:\Thesis\Main\Models\XGBoost\XGB_Final_Model_Features_name111.pkl')
print("✓ Feature names saved: feature_names.pkl")
```
